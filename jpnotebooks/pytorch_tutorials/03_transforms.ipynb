```python
# load data

import torch
import torch.nn.functional as F
from torchvision import datasets
from torchvision.transforms import v2

ds = datasets.FashionMNIST(
    root="data",
    train=True,
    download=True,
    # 2 step pipeline--> ToImage() converts PIL image to tensor, ToDtype converts tensor to float32 and normalises it to [0,1]
    transform=v2.Compose([v2.ToImage(), v2.ToDtype(torch.float32, scale=True)]),    # normalise to tensor
    target_transform=v2.Lambda(     # try out lambda-func (inline-func)
                # one hot is a tensor with 0's and one 1
        lambda y: F.one_hot(torch.tensor(y), num_classes=10).float()
    ),
)
```
